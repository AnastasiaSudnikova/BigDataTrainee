**4-5. Calculate the mean and standard deviation of the age of those who earn more than 50K per year (the *salary* attribute) and those who earn less than 50K per year**

```python
import pandas as pd

data = pd.read_csv("./data/adult.data.csv")

salary_gt_50k = data[data['salary'] == '>50K']
salary_le_50k = data[data['salary'] == '<=50K']

mean_age_gt_50k = salary_gt_50k['age'].mean()
std_age_gt_50k = salary_gt_50k['age'].std()

mean_age_le_50k = salary_le_50k['age'].mean()
std_age_le_50k = salary_le_50k['age'].std()

print("Статистика возраста по уровню дохода:")
print(f"Зарплата > 50K:")
print(f"  Средний возраст: {mean_age_gt_50k:.2f} лет")
print(f"  Стандартное отклонение: {std_age_gt_50k:.2f} лет")
print(f"  Количество: {len(salary_gt_50k)}")
print()
print(f"Зарплата <= 50K:")
print(f"  Средний возраст: {mean_age_le_50k:.2f} лет")
print(f"  Стандартное отклонение: {std_age_le_50k:.2f} лет")
print(f"  Количество: {len(salary_le_50k)}")
```

**6. Is it true that people who earn more than 50k have at least a higher education? (*education indicator – Bachelors, Prof-school, Assoc-acdm, Assoc-voc, Masters or Doctorate*)**

```python
import pandas as pd

data = pd.read_csv("./data/adult.data.csv")

higher_education = ['Bachelors', 'Prof-school', 'Assoc-acdm', 'Assoc-voc', 'Masters', 'Doctorate']

data['higher_education'] = data['education'].isin(higher_education)

education_salary = data.groupby(['salary', 'higher_education']).size().unstack()

higher_ed_percent = data.groupby('salary')['higher_education'].mean() * 100

print("Образование и доход:")

print("\nКоличество людей по образованию и доходу:")
print(education_salary)

print("\nПроцент людей с высшим образованием:")
print(higher_ed_percent.round(2))


gt_50k_percent = higher_ed_percent['>50K']
le_50k_percent = higher_ed_percent['<=50K']

print(f"Среди людей с доходом >50K: {gt_50k_percent:.1f}% имеют высшее образование")
print(f"Среди людей с доходом <=50K: {le_50k_percent:.1f}% имеют высшее образование")

if gt_50k_percent > le_50k_percent:
    print("\nДа, люди с доходом >50K чаще имеют высшее образование.")
    print(f"   Разница: {gt_50k_percent - le_50k_percent:.1f}%")
else:
    print("\nНет, люди с доходом >50K НЕ обязательно имеют высшее образование.")
```

**8. Among whom is the proportion of high-earning individuals (>50K) higher: married or single men (*marital-status* indicator)? We consider those whose *marital-status* begins with *Married* (Married-civ-spouse, Married-spouse-absent, or Married-AF-spouse) to be married; the rest are considered single.**

```python
import pandas as pd

data = pd.read_csv("./data/adult.data.csv")

men = data[data['sex'] == 'Male']

def marital_group(status):
    if status.startswith('Married'):
        return 'Married'
    else:
        return 'Single'

men['marital_group'] = men['marital-status'].apply(marital_group)

high_earner_by_marital = men.groupby('marital_group')['salary'].apply(
    lambda x: (x == '>50K').mean() * 100
)

print("Доля людей с доходом >50K среди мужчин по семейному статусу:")
print(high_earner_by_marital.round(2))

married_percent = high_earner_by_marital['Married']
single_percent = high_earner_by_marital['Single']

print(f"Среди женатых мужчин: {married_percent:.2f}% зарабатывают >50K")
print(f"Среди холостых мужчин: {single_percent:.2f}% зарабатывают >50K")

if married_percent > single_percent:
    print("\nСреди женатых мужчин доля высокооплачиваемых выше.")
    print(f"   Разница: {married_percent - single_percent:.2f}%")
else:
    print("\nСреди холостых мужчин доля высокооплачиваемых выше.")
    print(f"   Разница: {single_percent - married_percent:.2f}%")
```

**11.Group people into age groups *young*, *adult*, *retiree*, where:**
* *young* corresponds to 16-35 years old
* *adult* - 35-70 years old
* *retiree* - 70-100 years old

**Enter the name of the appropriate age group for each person in the new AgeGroup column.**

```python
import pandas as pd

data = pd.read_csv("./data/adult.data.csv")

def age_group(age):
    if age < 35:
        return 'young'
    elif age < 70:
        return 'adult'
    else:
        return 'retiree'

data['AgeGroup'] = data['age'].apply(age_group)

print(" Распределение по возрастным группам:")

print(data['AgeGroup'].value_counts())

print("\nПервые 5 строк с новой колонкой:")
print(data[['age', 'AgeGroup']].head())
```

**12-13. Determine the number of people earning >50K in each age group (AgeGroup column), and also output the name of the age group in which people most often earn more than 50K (>50K)**

```python
import pandas as pd

data = pd.read_csv("./data/adult.data.csv")

bins = [16, 35, 70, 100]
labels = ['young', 'adult', 'retiree']
data['AgeGroup'] = pd.cut(data['age'], bins=bins, labels=labels, right=False)

percent_high = data.groupby('AgeGroup')['salary'].apply(lambda x: (x == '>50K').mean() * 100).round(2)

best_group = percent_high.idxmax()
best_percent = percent_high.max()

print("Процент людей с доходом >50K по возрастным группам:")
print(percent_high)

print(f"Возрастная группа, где чаще всего зарабатывают >50K: {best_group}")
print(f"   Процент: {best_percent:.2f}%")
```

**14. Group people by employment type (the occupation column) and determine the number of people in each group. Then write a filter function, filter_func, that will return only those groups in which the average age (the age column) is 40 or less and in which all employees work more than 5 hours per week (the hours-per-week column).**

```python
import pandas as pd

data = pd.read_csv("./data/adult.data.csv")

occupation_counts = data['occupation'].value_counts()

print("Количество людей по профессиям:")

print(occupation_counts)

def filter_func(group):
    avg_age_condition = group['age'].mean() <= 40
    
    all_hours_condition = (group['hours-per-week'] > 5).all()
    
    return avg_age_condition and all_hours_condition

filtered_groups = data.groupby('occupation').filter(filter_func)

print("Профессии, удовлетворяющие условиям:")
print("   - средний возраст <= 40")
print("   - все сотрудники работают > 5 часов в неделю")


filtered_occupations = filtered_groups['occupation'].unique()
print(f"Найдено профессий: {len(filtered_occupations)}")
print(f"Профессии: {sorted(filtered_occupations)}")

print("\nСтатистика по отфильтрованным профессиям:")
for occ in sorted(filtered_occupations):
    group = data[data['occupation'] == occ]
    avg_age = group['age'].mean()
    min_hours = group['hours-per-week'].min()
    count = len(group)
    print(f"{occ:<25} | Средний возраст: {avg_age:.2f} | Мин. часы: {min_hours} | Кол-во: {count}")
```
