**3.** Construct the direct product of arrays (all combinations with each element). A two-dimensional array is provided as input.

```python
import numpy as np

a = np.array([
    [1, 2],
    [3, 4]
])

b = np.array([
    [5, 6],
    [7, 8]
])

a_repeated = np.repeat(a, len(b), axis = 0)
b_repeated = np.tile(b, (len(a), 1))

result = np.hstack([a_repeated, b_repeated])
print(result)
```

**4.** Given two arrays, A (8x3) and B (2x2), find rows in A that contain elements from each row in B, regardless of the order of the elements in B.

```python
import numpy as np

A = np.array([
    [1, 2, 3],
    [4, 5, 6],
    [1, 3, 2],
    [4, 6, 5],
    [7, 8, 9],
    [2, 1, 3],
    [5, 4, 6],
    [9, 7, 8]
])

B = np.array([
    [1, 2],
    [4, 5]
])

B_sets = [set(row) for row in B] #превращаем строки B в множества для удобства проверки подмножеств {1, 2}, {4, 5}
mask = np.zeros(len(A), dtype=bool)

for i, row in enumerate(A):
    row_set = set(row)
    for b_set in B_sets:
        if b_set.issubset(row_set):
            mask[i] = True
            break
        
result = A[mask]
print('Строки А, которые содержат элементы из В:')
print(result)
```

**5.** Given a 10x3 matrix, find rows of unequal values ​​(for example, the row [2,2,3] remains, the row [3,3,3] is deleted)

```python
import numpy as np

matrix = np.array([
    [1, 2, 3],
    [4, 4, 4],
    [5, 5, 6],
    [7, 7, 7],
    [8, 9, 8],
    [1, 1, 1],
    [2, 2, 3],
    [3, 4, 5],
    [6, 6, 6],
    [7, 8, 9]
])

mask = []

for row in matrix:
    if len(np.unique(row)) > 1:
        mask.append(True)
    else:
        mask.append(False)

result = matrix[mask]
print(result)
```

* __Task 1__: Calculate the product of nonzero elements on the diagonal of a rectangular matrix.  
 For example, for X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]]) answer is 3.

```python
def product_nonzero_diagonal_python(matrix):
    product = 1
    min_dim = min(len(matrix), len(matrix[0]))  
    
    for i in range(min_dim):
        elem = matrix[i][i]
        if elem != 0:
            product *= elem
    
    return product


X = [[1, 0, 1],
     [2, 0, 2],
     [3, 0, 3],
     [4, 4, 4]]

print(product_nonzero_diagonal_python(X)) 
```

```python
import numpy as np

def product_nonzero_diagonal_numpy(matrix):
    diag = np.diag(matrix)
    
    nonzero_diag = diag[diag != 0]
  
    if nonzero_diag.size > 0:
        return np.prod(nonzero_diag)
    else:
        return 1  

X = np.array([[1, 0, 1],
              [2, 0, 2],
              [3, 0, 3],
              [4, 4, 4]])

print(product_nonzero_diagonal_numpy(X))  
```

* __Task 2__: Two vectors x and y are given. Check if they define the same multiset.  
  For example, for x = np.array([1, 2, 2, 4]), y = np.array([4, 2, 1, 2]) answer is True.

```python
def same_multiset_python(x, y):
    if len(x) != len(y):
        return False
    
    return sorted(x) == sorted(y)

x = [1, 2, 2, 4]
y = [4, 2, 1, 2]

print(same_multiset_python(x, y)) 
```

```python
import numpy as np

def same_multiset_numpy(x, y):
    if len(x) != len(y):
        return False
    
    return np.array_equal(np.sort(x), np.sort(y))

x = np.array([1, 2, 2, 4])
y = np.array([4, 2, 1, 2])

print(same_multiset_numpy(x, y)) 
```

* __Task 3__: Find the maximum element in the vector x among the elements preceded by zero. 
 For example, for x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0]) answer is 5.

```python
def max_preceded_by_zero_python(x):
    max_val = None
    
    for i in range(1, len(x)):
        if x[i-1] == 0: 
            if max_val is None or x[i] > max_val:
                max_val = x[i]
    
    return max_val

x = [6, 2, 0, 3, 0, 0, 5, 7, 0]
print(max_preceded_by_zero_python(x))  
```

```python
import numpy as np

def max_preceded_by_zero_numpy(x):
    mask = np.concatenate([[False], x[:-1] == 0])
    
    candidates = x[mask]
    
    if candidates.size > 0:
        return np.max(candidates)
    else:
        return None

x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0])
print(max_preceded_by_zero_numpy(x)) 
```

* __Task 4__: Implement run-length encoding. For some vector x, it is necessary to return a tuple of two vectors of the same length. The first one contains numbers and the second one contains how many times they need to be repeated.  
 For example, for x = np.array([2, 2, 2, 3, 3, 3, 5]) answer is (np.array([2, 3, 5]), np.array([3, 3, 1])).

```python
def run_length_encoding_python(x):
    if not x:
        return [], []
    
    values = []
    counts = []
    
    current_val = x[0]
    current_count = 1
    
    for i in range(1, len(x)):
        if x[i] == current_val:
            current_count += 1
        else:
            values.append(current_val)
            counts.append(current_count)
            current_val = x[i]
            current_count = 1
    
    values.append(current_val)
    counts.append(current_count)
    
    return values, counts

x = [2, 2, 2, 3, 3, 3, 5]
values, counts = run_length_encoding_python(x)
print(values)  
print(counts) 
```

```python
import numpy as np

def run_length_encoding_numpy(x):
    if len(x) == 0:
        return np.array([]), np.array([])
    
    changes = np.concatenate([[True], x[1:] != x[:-1]])
    
    values = x[changes]
    
    change_indices = np.concatenate([np.where(changes)[0], [len(x)]])
 
    counts = np.diff(change_indices)
    
    return values, counts

x = np.array([2, 2, 2, 3, 3, 3, 5])
values, counts = run_length_encoding_numpy(x)
print(values)  
print(counts) 
```

* __Task 5__: Two samples of objects are given - X and Y. Calculate the matrix of Euclidean distances between objects. Compare with function scipy.spatial.distance.cdist by operating speed.

```python
import math

def euclidean_distance_matrix_python(X, Y):
    result = []
    
    for x in X:
        row = []
        for y in Y:
            dist = math.sqrt(sum((a - b) ** 2 for a, b in zip(x, y)))
            row.append(dist)
        result.append(row)
    
    return result

X = [[1, 2], [3, 4], [5, 6]]
Y = [[1, 1], [2, 2]]

dist_matrix = euclidean_distance_matrix_python(X, Y)
for row in dist_matrix:
    print([round(val, 2) for val in row])
```

```python
import numpy as np

def euclidean_distance_matrix_numpy(X, Y):
    X_sq = np.sum(X**2, axis=1, keepdims=True)  
    Y_sq = np.sum(Y**2, axis=1)  
    
    dist_sq = X_sq - 2 * np.dot(X, Y.T) + Y_sq
  
    return np.sqrt(np.maximum(dist_sq, 0))  

X = np.array([[1, 2], [3, 4], [5, 6]])
Y = np.array([[1, 1], [2, 2]])

dist_matrix = euclidean_distance_matrix_numpy(X, Y)
print(dist_matrix)
```

5. While the median shows that at least half of our values ​​are over 100 calories, it would be more impressive to show that a significant portion of the competitors have higher calorie counts than CrunchieMunchies.

Calculate various percentiles and print them until you find the lowest percentile that exceeds 60 calories. Store this value in the nth_percentile variable.

```python
import numpy as np

calorie_stats = np.loadtxt("./data/cereal.csv", delimiter=",")

percentiles = np.arange(0, 101, 5)  

nth_percentile = None

for p in percentiles:
    value = np.percentile(calorie_stats, p)
    print(f"Перцентиль {p}%: {value:.1f} калорий")
    
    if value > 60:
        nth_percentile = p
        break

print(f"\nСамый низкий перцентиль, превышающий 60 калорий: {nth_percentile}%")
print(f"Значение: {np.percentile(calorie_stats, nth_percentile):.1f} калорий")
```
